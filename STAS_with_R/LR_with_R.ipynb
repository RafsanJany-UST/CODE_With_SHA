```r
# Load required libraries
library(data.table)
library(caret)



# 1. Load the dataset
data <- fread("C:/Users/user/Documents/GitHub/CODE_With_SHA/STATS_with_Py/Train.csv")

# 2. Convert all character columns to factors (simple encoding)
cat_cols <- names(Filter(is.character, data))
data[, (cat_cols) := lapply(.SD, as.factor), .SDcols = cat_cols]

# 3. Set the target column
target <- "Age (years)"

# 4. Remove rows with missing values
data <- na.omit(data)

# 5. Convert "Blood Pressure (s/d)" into two numeric columns
bp_split <- tstrsplit(data$`Blood Pressure (s/d)`, "/", type.convert = TRUE)
data[, `Systolic BP` := bp_split[[1]]]
data[, `Diastolic BP` := bp_split[[2]]]

data[, `Blood Pressure (s/d)` := NULL]  # Drop the original string column

# 6. Split into features and target
X <- data[, !target, with = FALSE]
y <- data[[target]]

# 7. Train-test split
set.seed(123)
trainIndex <- createDataPartition(y, p = 0.8, list = FALSE)
X_train <- X[trainIndex, ]
X_test  <- X[-trainIndex, ]
y_train <- y[trainIndex]
y_test  <- y[-trainIndex]

# 8. Combine features and target for modeling
train_data <- cbind(X_train, y = y_train)
test_data  <- cbind(X_test, y = y_test)

# 9. Train linear regression model
model <- lm(y ~ ., data = train_data)

# 10. Predict on test data
predictions <- predict(model, newdata = test_data)

# 11. Evaluate model performance
rmse <- sqrt(mean((predictions - y_test)^2, na.rm = TRUE))
r2 <- 1 - sum((predictions - y_test)^2, na.rm = TRUE) / sum((y_test - mean(y_test))^2)

# 12. Output results
cat("RMSE:", rmse, "\n")
cat("R-squared:", r2, "\n")
```
